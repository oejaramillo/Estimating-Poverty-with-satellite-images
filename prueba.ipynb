```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os
import rasterio
import math
```

## Eyeball check

Recovering georeference from tiff for png

```python
png_path = "sentinel/quito_satellite_images_pruebas/uint8"
tiff_path = "sentinel/quito_satellite_images_pruebas/uint16"
images = os.listdir(path=png_path)
images.sort()
```

```python
def get_png_georeference(png_path, tiff_dir):
    """
    Given a PNG filename and the directory with corresponding TIFFs,
    recover the georeference (bounds + CRS + transform) from its TIFF pair.
    
    Parameters
    ----------
    png_path : str
        Path of the PNG file (e.g. /.../tile_0.png)
    tiff_dir : str
        Directory where GeoTIFFs are stored

    Returns
    -------
    dict
        Metadata with CRS, bounds, transform
    """

    # Match file by index
    base_name = os.path.basename(png_path)       # e.g. "tile_0.png"
    tile_id = os.path.splitext(base_name)[0]     # "tile_0"
    tiff_path = os.path.join(tiff_dir, f"{tile_id}.tiff")

    if not os.path.exists(tiff_path):
        raise FileNotFoundError(f"GeoTIFF not found for {png_path}")

    with rasterio.open(tiff_path) as src:
        metadata = {
            "crs": str(src.crs),
            "bounds": src.bounds,        # (minx, miny, maxx, maxy)
            "transform": src.transform,  # Affine transform
            "width": src.width,
            "height": src.height
        }

    return metadata
```

```python
def trunc(num, decimals=2):
    """Truncate a float to a fixed number of decimals (no rounding)."""
    factor = 10 ** decimals
    return math.trunc(num * factor) / factor
```

```python
for x in images:
    # --- Paths ---
    img_p = os.path.join(png_path, x)
    metadata = get_png_georeference(x, tiff_path)

    # --- Load + Adjust brightness ---
    img = mpimg.imread(img_p).astype(float)
    img = np.clip(img * 5, 0, 255)

    # --- Extract tile ID from filename (works for multi-digit tiles) ---
    tile_id = os.path.splitext(x)[0].split("_")[-1]

    # --- Plot ---
    plt.imshow(img)
    plt.axis("off")

    bounds = metadata['bounds']
    legend_text = (
        f"Tile {tile_id}\n"
        f"Bounds:\n"
        f" l:{trunc(bounds.left, 2)} "
        f" b:{trunc(bounds.bottom, 2)}\n"
        f" r:{trunc(bounds.right, 2)} "
        f" t:{trunc(bounds.top, 2)}"
    )

    plt.text(
        0.01, 0.99,
        legend_text,
        color="white",
        fontsize=10,
        ha="left", va="top",
        transform=plt.gca().transAxes,
        bbox=dict(facecolor="black", alpha=0.5)
    )

    # --- Save output ---
    out_file = os.path.join(
        "sentinel/quito_satellite_images_pruebas/png legned",
        f"tile_{tile_id}.png"
    )
    plt.savefig(out_file, bbox_inches="tight", pad_inches=0)
    plt.close()  # free memory for next tile

    print(f"Saved: {out_file}")
```
